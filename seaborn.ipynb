```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("laptopData.csv")
```

```python
# i wanna see outliiner using box plot for price column 

plt.figure(figsize=(7,5))
plt.title("Checking outline's")
sns.boxplot(df['Price'])
```

```python
try:
    sns.scatterplot(data=df ,
                x='Company',
                y='Price',
                hue ='TypeName',
                alpha=0.7)
    plt.title("Outlier's")
    plt.xlabel("Company named")
    plt.xticks(rotation=45,ha='right')
    plt.ylabel("Price")
    
except Exception as E:
    print("ERROR:",E)
```

```python
try:
    df['Price'] = df.groupby('TypeName')['Price'].transform(lambda x: x.fillna(x.median()))
    print(df['Price'].isnull().sum())
except Exception as E:
    print("ERROR: ",E)
```
