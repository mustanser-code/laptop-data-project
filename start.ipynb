```python
import pandas as pd
```

```python
# reading head of the raw data
df = pd.read_csv("laptopData.csv")
df.head()
```

```python
#first of all i wanna make the unnamed column into laptopn ID named
# Here how to do it 
df = df.rename(columns={"Unnamed: 0":'Laptop_ID'})
df.head()
```

```python
# getting information of raw data
df.info()
```

```python
# changing every column to desired datatype
# using build-in function
# i will use try-except to getting better error handling
try:
    df = df.convert_dtypes()
    display(df.dtypes)
    display(df.head())
except Exception as E:
    print("ERROR: ",E)
```

```python
# now checking null values inside my data set
df.isnull().sum()
```

```python
df['Laptop_ID'].isnull().sum()
```

```python
df.columns
```

```python
try:
    df['Company'].str.strip()

except Exception as E:
    display("Error: ",E)
```

```python
try:
    df['Gpu'] = df['Gpu'].str.replace(' ','',regex=False)
    display(df['Gpu'].head())
except Exception as E:
    display("ERROR: ", E)
```

```python
df.columns
```

```python
try:
    df["Ram"] = df["Ram"].str.strip()
    display(df["Ram"])
except Exception as E:
    print("ERROR",E)
```

```python
df["Price"].isnull().sum()
```

```python
df.columns
```

```python
try:
    df['Weight'] = df['Weight'].str.replace('kg',' ',case=False).str.strip()
except Exception as E:
    print("ERROR:" ,E)
```

```python
df['Weight'] = pd.to_numeric(df['Weight'],errors ='coerce')
df['Weight IN KG'] = df['Weight']
```

```python
df.to_csv("LAPTOPDATA.csv")
```
